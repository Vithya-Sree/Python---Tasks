1. List – Creation, Modification, and Access

```python
age_list = [24, 25, 27, 28, 29]
name_list = ["Anu", "Priya", "Kavi", "Divya", "Meena"]

print(age_list)
print(name_list)
```

```python
# Append
name_list.append("Yazhini")
print(name_list)
```

```python
#Insert
age_list.insert(2, 26)
print(age_list)
```

```python
#Remove
name_list.remove("Yazhini")
print(name_list)
```

```python
#Pop
print(age_list.pop())
print(age_list)
```

```python
#Extend
age_list.extend([31, 30, 32])
print(age_list)
```

```python
#Sort
age_list.sort(reverse=True)
print(age_list)
```

```python
#Find maximum, minimum, and sum
print("Maximum age : " ,max(age_list))
print("Minimum age : " ,min(age_list))
print("Sum of all ages : " ,sum(age_list))
```

Accessing List Elements:

```python
# Print the first element of name_list
print("First element of name_list:", name_list[0])
```

```python
# Print the last element of name_list
print("Last element of name_list:", name_list[-1])
```

```python
# Print elements from index 2 to index 4
print(name_list[2:5])
```

```python
# Print name_list in reverse order
print("list in reverse order:", name_list[::-1])
```

2. Dictionary – Creation, Modification, and Access

```python
# Create a dictionary containing five students and their marks
student_marks = {
    "Anu": 75,
    "Priya": 88,
    "Kavi": 92,
    "Divya": 69,
    "Meena": 78
}

print(student_marks)
```

```python
# Access and print the mark of a specific student
print("Mark of Kavi:", student_marks["Kavi"])
```

```python
# Add a new student, Janani
student_marks["Janani"] = 80
print(student_marks)
```

```python
# Update the mark of any one student to 82
student_marks["Anu"] = 82
print(student_marks)
```

```python
# Print all keys, values, and key-value pairs:
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Key-value pairs:", student_marks.items())
```

3. Sets – Operations

```python
# Create a set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

```python
# Attempt to access my_set[4]
print(my_set[4])

# Explanation: Sets are unordered collections and do not support indexing.
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

```python
# Union and intersection

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union of set1 and set2:", union_set)
print("Intersection of set1 and set2:", intersection_set)
```

Performance Category Program:

1. Prompt the user for input. Score range should be from 0 to 10 (both inclusive).
2. Find the performance category based on the input score using the following criteria: a. Above Average: Score greater than 7
b. Average: Score between 4 and 7 (both inclusive)
c. Below Average: Score less than 4
3. Output: Print the Performance category

```python
score = int(input("Enter your score (0 to 10): "))

if score < 0 and score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Performance: Above Average")
    print("Excellent performance! Keep up the great work.")

elif score >= 4:
    print("Performance: Average")
    print("Good effort! Keep practicing, there's room for improvement.")

else:
    print("Performance: Below Average")
    print("Need to improve your performance; consistent practice will lead to better results.")
```
