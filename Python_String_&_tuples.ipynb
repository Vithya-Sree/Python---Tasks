String Concatenation:

Write a Python program that takes two strings, i.e., string 1 "Hello" and string 2 “get name” as input from the user, and concatenates them together. Display the concatenated string as the output.

Sample Output:

Enter your Name: Zara

Hello Zara

```python
string1 = "Hello"
name = input("Enter Your Name: ")
```

```python
concat = string1 +" "+name
print(concat)
```

Tuples (Creation, Modification and Access)

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
```

```python
#a. Concatenate the two tuples and store it in “t_combine”
t_combine = tuple1 + tuple2
print("Concatenate : ",t_combine)
```

```python
#b. Repeat the elements of “t_combine” 3 times
print(t_combine * 3)
```

```python
#c. Access the 3rd element from “t_combine”
print("3rd element : ",t_combine[2])
```

```python
#d. Access the first three elements from “t_combine”
print("first three elements : ",t_combine[:3])
```

```python
#e. Access the last three elements from “t_combine”
print("Last three elements : ",t_combine[-3:])
```
